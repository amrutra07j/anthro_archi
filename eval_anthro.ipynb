````python
from anthropic import Anthropic
````

````python
client = Anthropic()
model = "claude-sonnet-5-5"
````

````python
import json

# Generate a dataset for prompt evaluation
def generate_dataset():
    prompt = """
Generate a evaluation dataset for a prompt evaluation. The dataset will be used to evaluate prompts
that generate Python, JSON, or Regex specifically for Pyspark-related tasks. Generate an array of JSON objects,
each representing task that requires Python, JSON, or a Regex to complete.

Example output:
```json
[
    {
        "task": "Description of task",
    },
    ...additional
]
```

* Focus on tasks that can be solved by writing a single Python function, a single JSON object, or a regular expression.
* Focus on tasks that do not require writing much code

Please generate 3 objects.
"""

    answers = client.messages.create(
        model=model,
        max_tokens=1024,
        messages=[{"role": "user", "content": prompt}],
        output_config={"format": {"type": "json_schema",
                                  "schema": {
                                      "type": "array",
                                      "items": {
                                          "type": "object",
                                            "properties": {
                                                "task": {"type": "string"},
                                                "format": {"type": "string", "enum": ["Python", "JSON", "Regex"]}
                                            },
                                            "additionalProperties": False,
                                      }
                                          
                                  }}}
    )
    return [block for block in answers.content if getattr(block, "type", None) == "text"][0].text
````

````python
datasets = generate_dataset()
datasets_json = json.loads(datasets)
````

````python
with open("datasets.json", "w") as f:
    json.dump(datasets_json, f, indent=4)
````
